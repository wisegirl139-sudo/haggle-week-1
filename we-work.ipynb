```python
import pandas as pd 
import numpy as np 
df= pd.read_csv("train.csv")
print(df.shape)
print(df.head())
print(df.columns.tolist())
```

```python
#now the maths and acc stuff 
y=np.log1p(df["hours_spent"])
X = df.drop(columns=["project_id", "hours_spent", "log_hours", "hours_bucket"])
print(X.shape)
links = X.corrwith(y, method="spearman", numeric_only=True) #spearman filters , filters is good bc now we know what to ignore and what to look out for way also i needed a list of names 
print(links.sort_values(key=abs, ascending=False).to_string()) ##finding useless stuff
```
